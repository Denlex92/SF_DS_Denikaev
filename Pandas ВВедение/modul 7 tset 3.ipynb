```python
import pandas as pd
import numpy as np

# Загрузка данных
diabetes = pd.read_csv('diabetes_data.csv')

print("="*60)
print("ЗАДАНИЕ 8.1: УДАЛЕНИЕ ДУБЛИКАТОВ")
print("="*60)

# Поиск дубликатов
duplicates = diabetes[diabetes.duplicated()]
print('Число дубликатов: {}'.format(duplicates.shape[0]))

# Удаление дубликатов
diabetes = diabetes.drop_duplicates()
print('Результирующее число записей: {}'.format(diabetes.shape[0]))

print("\n" + "="*60)
print("ЗАДАНИЕ 8.2: УДАЛЕНИЕ НЕИНФОРМАТИВНЫХ ПРИЗНАКОВ")
print("="*60)

def find_low_informative_feature(data):
    low_information_cols = []
    for col in data.columns:
        top_freq = data[col].value_counts(normalize=True).max()
        nunique_ratio = data[col].nunique() / data[col].count()
        if top_freq > 0.95:
            low_information_cols.append(col)
            print(f"❌ {col}: 95%+ значений одинаковы ({top_freq:.1%})")
        elif nunique_ratio > 0.95:
            low_information_cols.append(col)
            print(f"❌ {col}: 95%+ значений уникальны ({nunique_ratio:.1%})")
        else:
            print(f"✅ {col}: информативен")
    return low_information_cols

low_information_cols = find_low_informative_feature(diabetes)
print(f"\nНеинформативные признаки: {low_information_cols}")

# Удаление неинформативных признаков
diabetes = diabetes.drop(columns=low_information_cols)
print(f"Осталось признаков: {diabetes.shape[1]}")
print(f"Признаки: {list(diabetes.columns)}")

print("\n" + "="*60)
print("ЗАДАНИЕ 8.3: ЗАМЕНА СКРЫТЫХ ПРОПУСКОВ")
print("="*60)

# Столбцы, где 0 означает пропуск
columns_with_zero_missing = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Замена нулей на NaN:")
for column in columns_with_zero_missing:
    if column in diabetes.columns:
        zeros_before = (diabetes[column] == 0).sum()
        diabetes[column] = diabetes[column].replace(0, np.nan)
        nans_after = diabetes[column].isnull().sum()
        print(f"{column}: {zeros_before} нулей → {nans_after} NaN")

# Расчет доли пропусков в Insulin
if 'Insulin' in diabetes.columns:
    insulin_missing = diabetes['Insulin'].isnull().sum()
    insulin_total = len(diabetes)
    insulin_missing_ratio = insulin_missing / insulin_total
    insulin_missing_rounded = round(insulin_missing_ratio, 2)
    
    print(f"\nДоля пропусков в Insulin:")
    print(f"{insulin_missing} / {insulin_total} = {insulin_missing_ratio:.4f}")
    print(f"Округлено до сотых: {insulin_missing_rounded}")

print("\n" + "="*60)
print("ЗАДАНИЕ 8.4: УДАЛЕНИЕ ПРИЗНАКОВ С >30% ПРОПУСКОВ")
print("="*60)

# Расчет порога (70% заполненных значений)
thresh = diabetes.shape[0] * 0.7
print(f"Порог для сохранения признака: {thresh:.0f} заполненных значений")

# Удаление признаков с более чем 30% пропусков
diabetes = diabetes.dropna(thresh=thresh, axis=1)

print(f"Осталось признаков: {diabetes.shape[1]}")
print(f"Признаки: {list(diabetes.columns)}")

print("\n" + "="*60)
print("ЗАДАНИЕ 8.5: УДАЛЕНИЕ СТРОК С >2 ПРОПУСКАМИ")
print("="*60)

# Подсчет пропусков в каждой строке
missing_per_row = diabetes.isnull().sum(axis=1)

print("Распределение пропусков по строкам:")
for i in range(0, missing_per_row.max() + 1):
    count = (missing_per_row == i).sum()
    if count > 0:
        percentage = (count / len(diabetes)) * 100
        print(f"  {i} пропусков: {count} строк ({percentage:.1f}%)")

# Удаление строк с более чем 2 пропусками
rows_before = len(diabetes)
diabetes_final = diabetes[missing_per_row <= 2]
rows_after = len(diabetes_final)
rows_removed = rows_before - rows_after

print(f"\nРезультат удаления:")
print(f"Было строк: {rows_before}")
print(f"Удалено строк с >2 пропусками: {rows_removed}")
print(f"Осталось строк: {rows_after}")

# Проверка результата
max_missing_after = diabetes_final.isnull().sum(axis=1).max()
print(f"Максимальное количество пропусков после очистки: {max_missing_after}")

if max_missing_after <= 2:
    print("✓ Условие выполнено: нет строк с более чем 2 пропусками")

print("\n" + "="*60)
print("ФИНАЛЬНЫЕ ОТВЕТЫ")
print("="*60)

print("8.1 - Записей после удаления дубликатов:", diabetes.shape[0])
print("8.2 - Неинформативные признаки:", low_information_cols)
print("8.3 - Доля пропусков в Insulin:", insulin_missing_rounded)
print("8.4 - Признаков после удаления с >30% пропусков:", diabetes.shape[1])
print("8.5 - Записей после удаления строк с >2 пропусками:", rows_after)

print("\n" + "="*60)
print("ФИНАЛЬНАЯ СТАТИСТИКА")
print("="*60)

print(f"Размер конечных данных: {diabetes_final.shape}")
print(f"Признаки: {list(diabetes_final.columns)}")

# Сохранение очищенных данных
diabetes_final.to_csv('diabetes_cleaned_final.csv', index=False)
print("\n✓ Очищенные данные сохранены в 'diabetes_cleaned_final.csv'")
```

```python
print("\n" + "="*60)
print("ЗАДАНИЕ 8.6: ЗАМЕНА ПРОПУСКОВ НА МЕДИАНУ")
print("="*60)

# Сохраняем копию данных перед заменой пропусков
diabetes_before_imputation = diabetes_final.copy()

print("Пропуски перед заменой:")
for column in diabetes_final.columns:
    missing_count = diabetes_final[column].isnull().sum()
    if missing_count > 0:
        print(f"  {column}: {missing_count} пропусков")

# Заменяем пропуски на медиану для каждого столбца
for column in diabetes_final.columns:
    if diabetes_final[column].isnull().sum() > 0:
        median_value = diabetes_final[column].median()
        missing_before = diabetes_final[column].isnull().sum()
        diabetes_final[column] = diabetes_final[column].fillna(median_value)
        missing_after = diabetes_final[column].isnull().sum()
        print(f"  {column}: заменено {missing_before} пропусков на медиану {median_value:.2f}")

# Проверяем, что пропусков больше нет
total_missing_after = diabetes_final.isnull().sum().sum()
print(f"\nПропусков после замены: {total_missing_after}")

# Вычисляем среднее значение в столбце SkinThickness
if 'SkinThickness' in diabetes_final.columns:
    skin_thickness_mean = diabetes_final['SkinThickness'].mean()
    skin_thickness_mean_rounded = round(skin_thickness_mean, 1)
    
    print(f"\nСреднее значение в SkinThickness:")
    print(f"  Среднее: {skin_thickness_mean:.4f}")
    print(f"  Округлено до десятых: {skin_thickness_mean_rounded}")
    
    # Дополнительная статистика для проверки
    print(f"\nДополнительная статистика SkinThickness:")
    print(f"  Медиана: {diabetes_final['SkinThickness'].median():.2f}")
    print(f"  Минимум: {diabetes_final['SkinThickness'].min():.2f}")
    print(f"  Максимум: {diabetes_final['SkinThickness'].max():.2f}")
    print(f"  Стандартное отклонение: {diabetes_final['SkinThickness'].std():.2f}")
else:
    print("Столбец SkinThickness не найден в данных")

print("\n" + "="*60)
print("ФИНАЛЬНЫЕ ОТВЕТЫ ДЛЯ 8.6")
print("="*60)

if 'SkinThickness' in diabetes_final.columns:
    print(f"8.6 - Среднее значение в SkinThickness: {skin_thickness_mean_rounded}")
else:
    print("8.6 - Столбец SkinThickness отсутствует")

# Сохранение данных после импутации
diabetes_final.to_csv('diabetes_cleaned_imputed.csv', index=False)
print("\n✓ Данные после замены пропусков сохранены в 'diabetes_cleaned_imputed.csv'")
```

```python
print("\n" + "="*60)
print("ЗАДАНИЕ 8.7: ПОИСК ВЫБРОСОВ В SKINTHICKNESS (IQR МЕТОД)")
print("="*60)

if 'SkinThickness' in diabetes_final.columns:
    # Берем данные из SkinThickness
    skin_data = diabetes_final['SkinThickness']
    
    # Вычисляем квартили и IQR
    Q1 = skin_data.quantile(0.25)
    Q3 = skin_data.quantile(0.75)
    IQR = Q3 - Q1
    
    # Вычисляем границы для выбросов
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    print(f"Статистика SkinThickness:")
    print(f"  Q1 (25-й перцентиль): {Q1:.2f}")
    print(f"  Q3 (75-й перцентиль): {Q3:.2f}")
    print(f"  IQR (межквартильный размах): {IQR:.2f}")
    print(f"  Нижняя граница выбросов: {lower_bound:.2f}")
    print(f"  Верхняя граница выбросов: {upper_bound:.2f}")
    print(f"  Минимум: {skin_data.min():.2f}")
    print(f"  Максимум: {skin_data.max():.2f}")
    
    # Находим выбросы
    outliers_lower = skin_data[skin_data < lower_bound]
    outliers_upper = skin_data[skin_data > upper_bound]
    total_outliers = len(outliers_lower) + len(outliers_upper)
    
    print(f"\nВыбросы:")
    print(f"  Выбросы ниже {lower_bound:.2f}: {len(outliers_lower)}")
    print(f"  Выбросы выше {upper_bound:.2f}: {len(outliers_upper)}")
    print(f"  Всего выбросов: {total_outliers}")
    
    # Показываем примеры выбросов
    if len(outliers_lower) > 0:
        print(f"  Примеры нижних выбросов: {outliers_lower.head().values}")
    if len(outliers_upper) > 0:
        print(f"  Примеры верхних выбросов: {outliers_upper.head().values}")
    
    # Визуализация для наглядности
    print(f"\nРаспределение значений:")
    print(f"  Значения в пределах [Q1-1.5*IQR, Q3+1.5*IQR]: {len(skin_data[(skin_data >= lower_bound) & (skin_data <= upper_bound)])}")
    print(f"  Значения за пределами: {total_outliers}")
    
    print("\n" + "="*60)
    print("ОТВЕТ ДЛЯ ЗАДАНИЯ 8.7")
    print("="*60)
    print(f"Количество выбросов в SkinThickness: {total_outliers}")

else:
    print("Столбец SkinThickness не найден в данных")
```

```python
print("\n" + "="*60)
print("ЗАДАНИЕ 8.8: ПОИСК ВЫБРОСОВ В SKINTHICKNESS (Z-SCORE МЕТОД)")
print("="*60)

if 'SkinThickness' in diabetes_final.columns:
    # Берем данные из SkinThickness
    skin_data = diabetes_final['SkinThickness']
    
    # Вычисляем среднее и стандартное отклонение
    mean_skin = skin_data.mean()
    std_skin = skin_data.std()
    
    # Вычисляем z-score для каждого значения
    z_scores = (skin_data - mean_skin) / std_skin
    
    # Находим выбросы (обычно |z-score| > 2 или > 3)
    # Используем стандартный порог |z| > 3 для сильных выбросов
    outliers_z = skin_data[np.abs(z_scores) > 3]
    
    print(f"Статистика SkinThickness:")
    print(f"  Среднее: {mean_skin:.2f}")
    print(f"  Стандартное отклонение: {std_skin:.2f}")
    print(f"  Минимум: {skin_data.min():.2f}")
    print(f"  Максимум: {skin_data.max():.2f}")
    
    print(f"\nZ-score анализ:")
    print(f"  Минимальный z-score: {z_scores.min():.2f}")
    print(f"  Максимальный z-score: {z_scores.max():.2f}")
    print(f"  Количество |z-score| > 3: {len(outliers_z)}")
    
    # Детальный анализ выбросов по z-score
    print(f"\nДетали выбросов (|z-score| > 3):")
    if len(outliers_z) > 0:
        for value in outliers_z.unique():
            z_score = (value - mean_skin) / std_skin
            count = (skin_data == value).sum()
            print(f"  Значение {value}: z-score = {z_score:.2f}, количество = {count}")
    else:
        print("  Выбросов не найдено")
    
    # Сравнение с разными порогами z-score
    print(f"\nСравнение разных порогов z-score:")
    for threshold in [2, 2.5, 3]:
        outliers_count = len(skin_data[np.abs(z_scores) > threshold])
        print(f"  |z-score| > {threshold}: {outliers_count} выбросов")
    
    print("\n" + "="*60)
    print("ОТВЕТ ДЛЯ ЗАДАНИЯ 8.8")
    print("="*60)
    print(f"Количество выбросов в SkinThickness (z-score > 3): {len(outliers_z)}")

else:
    print("Столбец SkinThickness не найден в данных")
```

```python
print("\n" + "="*60)
print("ЗАДАНИЕ: ВЫБРОСЫ В DIABETESPEDIGREEFUNCTION")
print("="*60)

def count_outliers_iqr(data_series):
    """Подсчет выбросов по методу IQR"""
    Q1 = data_series.quantile(0.25)
    Q3 = data_series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data_series[(data_series < lower_bound) | (data_series > upper_bound)]
    return len(outliers), Q1, Q3, IQR, lower_bound, upper_bound

# Анализ исходного признака
if 'DiabetesPedigreeFunction' in diabetes_final.columns:
    original_data = diabetes_final['DiabetesPedigreeFunction']
    
    # Выбросы в исходных данных
    num_outliers_original, Q1_orig, Q3_orig, IQR_orig, lower_orig, upper_orig = count_outliers_iqr(original_data)
    
    print("ИСХОДНЫЙ ПРИЗНАК:")
    print(f"Q1 = {Q1_orig:.3f}")
    print(f"Q3 = {Q3_orig:.3f}")
    print(f"IQR = {IQR_orig:.3f}")
    print(f"Нижняя граница = {lower_orig:.3f}")
    print(f"Верхняя граница = {upper_orig:.3f}")
    print(f"Число выбросов: {num_outliers_original}")
    
    # Выбросы в логарифмированном признаке (без +1)
    log_data = np.log(original_data)
    num_outliers_log, Q1_log, Q3_log, IQR_log, lower_log, upper_log = count_outliers_iqr(log_data)
    
    print("\nЛОГАРИФМИРОВАННЫЙ ПРИЗНАК:")
    print(f"Q1 = {Q1_log:.3f}")
    print(f"Q3 = {Q3_log:.3f}")
    print(f"IQR = {IQR_log:.3f}")
    print(f"Нижняя граница = {lower_log:.3f}")
    print(f"Верхняя граница = {upper_log:.3f}")
    print(f"Число выбросов: {num_outliers_log}")
    
    # Разница
    difference = num_outliers_original - num_outliers_log
    
    print(f"\nРАЗНИЦА: {num_outliers_original} - {num_outliers_log} = {difference}")
    
    # Визуализация для наглядности
    print(f"\nСтатистика исходного признака:")
    print(f"Минимум: {original_data.min():.3f}")
    print(f"Максимум: {original_data.max():.3f}")
    print(f"Медиана: {original_data.median():.3f}")
    print(f"Среднее: {original_data.mean():.3f}")
    
    print(f"\nСтатистика логарифмированного признака:")
    print(f"Минимум: {log_data.min():.3f}")
    print(f"Максимум: {log_data.max():.3f}")
    print(f"Медиана: {log_data.median():.3f}")
    print(f"Среднее: {log_data.mean():.3f}")
    
else:
    print("Признак DiabetesPedigreeFunction отсутствует в данных!")
```
