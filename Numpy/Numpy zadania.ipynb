Напишите функцию get_chess(a), которая принимает на вход длину стороны квадрата a и возвращает двумерный массив формы (a, a), заполненный 0 и 1 в шахматном порядке. В левом верхнем углу всегда должен быть ноль.

Примечание. воспользуйтесь функцией np.zeros, а затем с помощью срезов без циклов задайте необходимым элементам значение 1.

В Python для получения каждого второго элемента используется срез [::2]. Подумайте, как грамотно применить этот принцип к двумерному массиву.

```python
import numpy as np
def get_chess(a):
    zeros = np.zeros((a, a))
    zeros[1::2,::2]=1
    zeros[::2,1::2]=1
    
    return zeros
    

print(get_chess(2))
```

Пользователю может настолько понравиться перемешанная версия плейлиста, что он захочет сохранить его копию. 
Однако вы не хотите хранить в памяти новую версию плейлиста, а просто хотите сохранять тот seed, с которым он был сгенерирован.

Для этого напишите функцию shuffle_seed(array), которая принимает на вход массив из чисел, генерирует случайное число для seed в диапазоне от 0 до 2**32 - 1 (включительно) и возвращает кортеж: перемешанный с данным seed массив (исходный массив должен оставаться без изменений), а также seed, с которым этот массив был получен.

Пример


array = [1, 2, 3, 4, 5]
shuffle_seed(array)
# (array([1, 3, 2, 4, 5]), 2332342819)
shuffle_seed(array)
# (array([4, 5, 2, 3, 1]), 4155165971)

```python
import numpy as np

def shuffle_seed(array):
    seed_value = np.random.randint(1,2**32 - 1 )
    array_shuffle = np.array(array).copy()
    np.random.seed(seed_value)
    np.random.shuffle(array_shuffle)
    
    return(array_shuffle, seed_value)


#def shuffle_seed(array):
    # Шаг 1: Генерируем случайный seed
    seed = np.random.randint(0, 2**32)  # например, 123456789
    
    # Шаг 2: Создаем копию массива
    array_copy = np.array(array).copy()  # [1, 2, 3, 4, 5]
    
    # Шаг 3: Устанавливаем seed
    np.random.seed(seed)  # "Открываем книгу на странице 123456789"
    
    # Шаг 4: Перемешиваем копию
    np.random.shuffle(array_copy)  # Читаем числа с этой страницы для перемешивания
    
    return (array_copy, seed)
```

Напишите функцию min_max_dist(*vectors), которая принимает на вход неограниченное число векторов через запятую. Гарантируется, что все векторы, которые передаются, одинаковой длины.

Функция возвращает минимальное и максимальное расстояние между векторами в виде кортежа.

Пример


vec1 = np.array([1,2,3])
vec2 = np.array([4,5,6])
vec3 = np.array([7, 8, 9])

min_max_dist(vec1, vec2, vec3)
# (5.196152422706632, 10.392304845413264)

Объяснение работы:

*vectors - принимает любое количество аргументов

np.array(vectors) - преобразует входные векторы в массив NumPy

Двойной цикл - перебирает все уникальные пары векторов (i, j где i < j)

np.linalg.norm(vectors[i] - vectors[j]) - вычисляет Евклидово расстояние:

vectors[i] - vectors[j] - разность векторов

np.linalg.norm() - вычисляет длину вектора (корень из суммы квадратов)

Обновление min/max - отслеживаем минимальное и максимальное расстояние

Математическая формула расстояния:
Для векторов a = [a₁, a₂, a₃] и b = [b₁, b₂, b₃]:

text
distance = √((a₁-b₁)² + (a₂-b₂)² + (a₃-b₃)²)

```python
import numpy as np
def min_max_dist(*vectors):
    vectors = np.array(vectors)
    min_dist = float('inf')
    max_dist = 0
    n = len(vectors)
    for i in range(n):
        for j in range(i+1,n):
            dist = np.linalg.norm(vectors[i] - vectors[j])
            if dist < min_dist:
                min_dist = dist
            if dist > max_dist:
                max_dist = dist
                
    return(min_dist,max_dist)
```

Напишите функцию any_normal, которая принимает на вход неограниченное число векторов через запятую. Гарантируется, что все векторы, которые передаются, одинаковой длины.

Функция возвращает True, если есть хотя бы одна пара перпендикулярных векторов. Иначе возвращает False.

```python
import numpy as np
def any_normal(*vectors):
    vectors = np.array(vectors)
    n = len(vectors)
    for i in range(n):
        for j in range(i+1,n):
            check = np.dot(i,j)
            if check == 0:
                return True
            else:
                return False
```

Объяснение ключевых моментов:

np.random.choice(range(1, 101), size=25, replace=False) - выбирает 25 уникальных чисел из диапазона 1-100

replace=False гарантирует, что числа не повторяются

reshape(5, 5) - преобразует линейный массив из 25 чисел в матрицу 5×5

Важно: Числа уникальны только в пределах одного поля, но могут повторяться между разными полями

```python
def get_unique_loto(num):
    # Генерируем все числа сразу
    all_numbers = np.random.choice(range(1, 101), size=(num, 25), replace=False)
    # Преобразуем в поля 5x5
    return all_numbers.reshape(num, 5, 5)
```
